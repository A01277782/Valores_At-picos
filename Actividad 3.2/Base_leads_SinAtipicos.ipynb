```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df= pd.read_csv('Base_de_Leads_SinNulos.csv') 
```

```python
df.info()
```

```python
#Se elimina la columna 'Unnamed: 0' que es un índice innecesario
df.drop(columns=['Unnamed: 0'], inplace=True)
```

```python
valores_nulos=df.isnull().sum()
valores_nulos
```

```python
# Creo 2 dataframes para poder procesar los outliers
# Variable cuantitativa
cuantitativas = df.iloc[:, 1:2]
# Variables cualitativas
cualitativas = pd.concat([df.iloc[:, 0:1], df.iloc[:, 2:12]], axis=1)
```

```python
#Realizamos diagrama de caja o bigote de cada columna del dataframe
fig = plt.figure(figsize =(15, 8))
cuantitativas.plot(kind='box', vert=False)
plt.title("Valores Atípicos del Dataframe")
plt.show() #dibujamos el diagrama
```

No se realizaron cambios en los datos, ya que después de analizar las variables cuantitativas no se identificaron valores atípicos. Por lo tanto, no fue necesario eliminar ni modificar ninguna observación, manteniendo la información original del conjunto de datos.
